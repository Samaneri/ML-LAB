```python
import pandas as pd
from sklearn.datasets import load_iris
from sklearn.svm import SVC
from sklearn.model_selection import train_test_split
from sklearn.metrics import confusion_matrix
from sklearn.metrics import classification_report
from sklearn.metrics import accuracy_score
iris = load_iris()
X = iris.data
y = iris.target
df = pd.DataFrame(X, columns=iris.feature_names)
df["target"] = y
print("First 10 records of Iris Dataset:")
print(df.head(10)
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=0
)
model = SVC(kernel='linear', random_state=0)
model.fit(X_train, y_train)
pred = model.predict(X_test)
print("\nConfusion Matrix for Linear Kernel:")
print(confusion_matrix(y_test, pred))
print("\nAccuracy for Linear Kernel:")
print(accuracy_score(y_test, pred))
print("\nClassification Report for Linear Kernel:")
print(classification_report(y_test, pred))
model1 = SVC(kernel='rbf', random_state=0)
model1.fit(X_train, y_train)
pred1 = model1.predict(X_test)
print("\nConfusion Matrix for RBF Kernel:")
print(confusion_matrix(y_test, pred1))
print("\nAccuracy for RBF Kernel:")
print(accuracy_score(y_test, pred1))
print("\nClassification Report for RBF Kernel:")
print(classification_report(y_test, pred1))
```
