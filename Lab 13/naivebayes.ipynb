```python
data = [
    ["1", ">=9", "Yes", "Very Good", "Good", "Yes"],
    ["2", ">=8", "No", "Good", "Moderate", "Yes"],
    ["3", ">=9", "No", "Average", "Poor", "No"],
    ["4", "<8", "No", "Average", "Good", "No"],
    ["5", ">=8", "Yes", "Good", "Moderate", "Yes"],
    ["6", ">=9", "Yes", "Good", "Moderate", "Yes"],
    ["7", "<8", "Yes", "Good", "Poor", "No"],
    ["8", ">=9", "No", "Good", "Good", "Yes"],
    ["9", ">=8", "Yes", "Good", "Good", "Yes"],
    ["10", ">=8", "Yes", "Average", "Good", "Yes"]
]
features = [
    "CGPA",
    "Interactiveness",
    "Practical Skills",
    "Communication"
]
test = {
    "CGPA": ">=9",
    "Interactiveness": "Yes",
    "Practical Skills": "Average",
    "Communication": "Good"
}
yes_data = [row for row in data if row[5] == "Yes"]
no_data = [row for row in data if row[5] == "No"]
print("Total records:", len(data))
print("Job Offer = Yes:", len(yes_data))
print("Job Offer = No :", len(no_data))
p_yes = len(yes_data) / len(data)
p_no = len(no_data) / len(data)
print("\nPRIOR PROBABILITY")
print("P(Yes) =", p_yes)
print("P(No)  =", p_no)
print("\nFREQUENCY MATRIX")
frequency = {}
for i, feature in enumerate(features, start=1):
    frequency[feature] = {}
    values = sorted(set(row[i] for row in data))
    print("\n", feature)
    print("Value\t\tYes\tNo")
    for value in values:
        yes_count = sum(1 for row in yes_data if row[i] == value)
        no_count = sum(1 for row in no_data if row[i] == value)
        frequency[feature][value] = {
            "Yes": yes_count,
            "No": no_count
        }
        print(f"{value:<12}\t{yes_count}\t{no_count}")
print("\nLIKELIHOOD PROBABILITY MATRIX")
likelihood = {}
for i, feature in enumerate(features, start=1):
    likelihood[feature] = {}
    values = sorted(set(row[i] for row in data))
    print("\n", feature)
    print("Value\t\tP(Value|Yes)\tP(Value|No)")
    for value in values:
        yes_count = frequency[feature][value]["Yes"]
        no_count = frequency[feature][value]["No"]
        p_value_yes = yes_count / len(yes_data)
        p_value_no = no_count / len(no_data)
        likelihood[feature][value] = {
            "Yes": p_value_yes,
            "No": p_value_no
        }
        print(
            f"{value:<12}\t"
            f"{p_value_yes:.4f}\t\t"
            f"{p_value_no:.4f}"
        )
print("\nNAIVE BAYES CALCULATION")
yes_score = p_yes
no_score = p_no
print("\nFor Job Offer = YES")
for feature in features:
    probability = likelihood[feature][test[feature]]["Yes"]
    print(
        f"P({test[feature]} | Yes) = "
        f"{probability:.4f}"
    )
    yes_score *= probability
print("\nFor Job Offer = NO")
for feature in features:
    probability = likelihood[feature][test[feature]]["No"]
    print(
        f"P({test[feature]} | No) = "
        f"{probability:.4f}"
    )
    no_score *= probability

print("\nFINAL RESULT")
print("Naive Bayes Score for Yes =", yes_score)
print("Naive Bayes Score for No  =", no_score)
if yes_score > no_score:
    prediction = "Yes"
else:
    prediction = "No"
print("\nStudent Details:")
print("CGPA              :", test["CGPA"])
print("Interactiveness   :", test["Interactiveness"])
print("Practical Skills  :", test["Practical Skills"])
print("Communication     :", test["Communication"])
print("\nFinal Prediction:", prediction)
if prediction == "Yes":
    print("The student is predicted to receive a JOB OFFER.")
else:
    print("The student is predicted NOT to receive a JOB OFFER.")
```

```python
data = [
    ["1", ">=9", "Yes", "Very Good", "Good", "Yes"],
    ["2", ">=8", "No", "Good", "Moderate", "Yes"],
    ["3", ">=9", "No", "Average", "Poor", "No"],
    ["4", "<8", "No", "Average", "Good", "No"],
    ["5", ">=8", "Yes", "Good", "Moderate", "Yes"],
    ["6", ">=9", "Yes", "Good", "Moderate", "Yes"],
    ["7", "<8", "Yes", "Good", "Poor", "No"],
    ["8", ">=9", "No", "Good", "Good", "Yes"],
    ["9", ">=8", "Yes", "Good", "Good", "Yes"],
    ["10", ">=8", "Yes", "Average", "Good", "Yes"]
]
features = [
    "CGPA",
    "Interactiveness",
    "Practical Skills",
    "Communication"
]
test = {
    "CGPA": ">=9",
    "Interactiveness": "Yes",
    "Practical Skills": "Average",
    "Communication": "Good"
}
yes_data = [row for row in data if row[5] == "Yes"]
no_data = [row for row in data if row[5] == "No"]
print("NAIVE BAYES JOB OFFER PREDICTION")
print("\nTotal Records       :", len(data))
print("Job Offer = Yes     :", len(yes_data))
print("Job Offer = No      :", len(no_data))
p_yes = len(yes_data) / len(data)
p_no = len(no_data) / len(data)
print("PRIOR PROBABILITY")
print("P(Yes) =", p_yes)
print("P(No)  =", p_no)
print("FREQUENCY MATRIX")
frequency = {}
for i, feature in enumerate(features, start=1):
    frequency[feature] = {}
    values = list(dict.fromkeys(row[i] for row in data))
    print("\n" + feature)
    print("Value\t\tYes\tNo")
    for value in values:
        yes_count = sum(
            1 for row in yes_data
            if row[i] == value
        )
        no_count = sum(
            1 for row in no_data
            if row[i] == value
        )
        frequency[feature][value] = {
            "Yes": yes_count,
            "No": no_count
        }
        print(f"{value:<12}\t{yes_count}\t{no_count}")
print("LIKELIHOOD PROBABILITY MATRIX")
likelihood = {}
for i, feature in enumerate(features, start=1):
    likelihood[feature] = {}
    values = list(dict.fromkeys(row[i] for row in data))
    print("\n" + feature)
    print("Value\t\tP(Value|Yes)\tP(Value|No)")
    for value in values:
        yes_count = frequency[feature][value]["Yes"]
        no_count = frequency[feature][value]["No"]
        p_value_yes = yes_count / len(yes_data)
        p_value_no = no_count / len(no_data)
        likelihood[feature][value] = {
            "Yes": p_value_yes,
            "No": p_value_no
        }
        print(
            f"{value:<12}\t"
            f"{p_value_yes:.4f}\t\t"
            f"{p_value_no:.4f}"
        )
print("NAIVE BAYES CALCULATION")
print("\nTest Student:")
print("CGPA              :", test["CGPA"])
print("Interactiveness   :", test["Interactiveness"])
print("Practical Skills  :", test["Practical Skills"])
print("Communication     :", test["Communication"])
yes_score = p_yes
print("\n--- For Job Offer = YES ---")
for feature in features:
    probability = likelihood[feature][test[feature]]["Yes"]
    print(
        f"P({test[feature]} | Yes) = "
        f"{probability:.4f}"
    )
    yes_score *= probability
no_score = p_no
print("\n--- For Job Offer = NO ---")
for feature in features:
    probability = likelihood[feature][test[feature]]["No"]
    print(
        f"P({test[feature]} | No) = "
        f"{probability:.4f}"
    )
    no_score *= probability
print("FINAL RESULT")
print("Naive Bayes Score for Yes =", yes_score)
print("Naive Bayes Score for No  =", no_score)
if yes_score > no_score:
    prediction = "Yes"
else:
    prediction = "No"
print("\nFinal Prediction:", prediction)
if prediction == "Yes":
    print("The student is predicted to receive a JOB OFFER.")
else:
    print("The student is predicted NOT to receive a JOB OFFER.")
print("CONFUSION MATRIX")
actual = []
predicted = []
for row in data:
    student_test = {
        "CGPA": row[1],
        "Interactiveness": row[2],
        "Practical Skills": row[3],
        "Communication": row[4]
    }
    student_yes_score = p_yes
    student_no_score = p_no
    for feature in features:
        yes_probability = likelihood[
            feature
        ][
            student_test[feature]
        ]["Yes"]

        no_probability = likelihood[
            feature
        ][
            student_test[feature]
        ]["No"]

        student_yes_score *= yes_probability
        student_no_score *= no_probability
    if student_yes_score > student_no_score:
        student_prediction = "Yes"
    else:
        student_prediction = "No"
    actual.append(row[5])
    predicted.append(student_prediction)
    print(
        row[0],
        "Actual =", row[5],
        "Predicted =", student_prediction
    )
TP = 0
TN = 0
FP = 0
FN = 0
for a, p in zip(actual, predicted):
    if a == "Yes" and p == "Yes":
        TP += 1
    elif a == "No" and p == "No":
        TN += 1
    elif a == "No" and p == "Yes":
        FP += 1
    elif a == "Yes" and p == "No":
        FN += 1
print("              CONFUSION MATRIX VALUES")
print("\n                 Predicted")
print("                 Yes    No")
print("Actual Yes       ", TP, "    ", FN)
print("Actual No        ", FP, "    ", TN)
print("\nTrue Positive  (TP) =", TP)
print("True Negative  (TN) =", TN)
print("False Positive (FP) =", FP)
print("False Negative (FN) =", FN)
total = TP + TN + FP + FN
accuracy = (TP + TN) / total
print("ACCURACY")
print("Accuracy =", accuracy)
print("Accuracy =", accuracy * 100, "%")
```
