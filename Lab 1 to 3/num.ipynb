```python
import numpy as np
print("NumPy Version:", np.__version__)
```

```python
x = np.array([1, 2, 3, 4, 5])
print("\nOriginal Array:")
print(x)
```

```python
print("Type of x:")
print(type(x))
```

```python
print("\nShape of x:")
print(x.shape)
```

```python
print("\nArray using arange():")
a = np.arange(1, 10, 5)
print(a)
print("\nArray using linspace():")
b = np.linspace(1, 10, 5)
print(b)
```

```python
x = np.array([1, 2, 3, 4, 5, 6])
print("\nOriginal Array:")
print(x)
print("\nReshape (2 x 3):")
print(x.reshape(2, 3))
```

```python
print("\nSize:")
print(x.size)
```

```python
print("\nShape:")
print(x.shape)
```

```python
print("\nData Type:")
print(x.dtype)
```

```python
print("\nItem Size (bytes):")
print(x.itemsize)
print("\nZero Array (4 x 5):")
zeros_array = np.zeros((4, 5))
print(zeros_array)
print("\nOnes Array (4 x 5):")
ones_array = np.ones((4, 5))
print(ones_array)
```

```python
print("\nRandom Array (4 x 5):")
random_array = np.random.random((4, 5))
print(random_array)
print("\nRandom Integer Array:")
rand_int = np.random.randint(0, 10, 5)
print(rand_int)
print("\nBefore Shuffle:")
print(rand_int)
np.random.shuffle(rand_int)
print("After Shuffle:")
print(rand_int)
```

```python
print("\nRandom Choice from Array:")
choice = np.random.choice(rand_int)
print(choice)
```
