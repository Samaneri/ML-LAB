```python
from matplotlib import pyplot as plt
import numpy as np

x = np.linspace(-np.pi, np.pi, 100)
y = np.sin(x)

plt.plot(x, y)
plt.xlabel("Angle")
plt.ylabel("Sine Function")
plt.title("Sine Wave")
plt.show()
```

```python
from matplotlib import pyplot as plt
import numpy as np

x = np.linspace(0, 10, 100)
y = x**2

plt.plot(x, y, 'g')
plt.xlabel("X")
plt.ylabel("Y")
plt.title("y = x²")
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

data = np.random.randn(100)

plt.hist(data, bins=10)
plt.title("Histogram")
plt.show()
```

```python
import matplotlib.pyplot as plt

marks = [40, 30, 20, 10]
labels = ["Python", "Java", "C", "C++"]

plt.pie(marks, labels=labels, autopct="%1.1f%%")
plt.title("Pie Chart")
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.linspace(0, 5, 100)

plt.subplot(211)
plt.plot(x, x**2)
plt.title("First Plot")

plt.subplot(212)
plt.plot(x, x**3)
plt.title("Second Plot")

plt.tight_layout()
plt.show()
```
