```python
import numpy as np
from sklearn import preprocessing

input_data = np.array([
    [0.2,0.8],
    [0.6,0.4]
])

output = preprocessing.Binarizer(threshold=0.6).transform(input_data)

print(output)
```

```python
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsClassifier

iris = load_iris()

X_train, X_test, y_train, y_test = train_test_split(
    iris.data, iris.target, test_size=0.3, random_state=1)

classifier = KNeighborsClassifier(n_neighbors=4)

classifier.fit(X_train, y_train)

print(classifier.predict(X_test[:5]))
```

```python
from sklearn.datasets import load_iris
from sklearn.neighbors import KNeighborsClassifier
import joblib

iris = load_iris()

model = KNeighborsClassifier()

model.fit(iris.data, iris.target)

joblib.dump(model, "sample_model.joblib")

print("Model Saved")
```

```python
import joblib

model = joblib.load("sample_model.joblib")

print("Model Loaded Successfully")
```
