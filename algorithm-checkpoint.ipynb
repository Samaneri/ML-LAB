```python
import pandas as pd

data = [
    ['Maths', 'High', 'Yes', 'Yes'],
    ['Maths', 'High', 'No', 'Yes'],
    ['Science', 'Low', 'Yes', 'No'],
    ['Maths', 'High', 'Yes', 'Yes']
]


hypothesis = data[0][:-1]

print("I1 =", data[0][:-1])
print("H1 =", hypothesis)

count = 2


for i in range(1, len(data)):

    print(f"I{i+1} =", data[i][:-1])

    if data[i][-1] == 'Yes':

        for j in range(len(hypothesis)):

            if hypothesis[j] != data[i][j]:
                hypothesis[j] = '?'

    print(f"H{count} =", hypothesis)

    count += 1

print("Final Hypothesis =", hypothesis)
```

```python
import pandas as pd
data = [
    ['Maths', 'High', 'Yes', 'Yes'],
    ['Maths', 'High', 'No', 'Yes'],
    ['Science', 'Low', 'Yes', 'No'],
    ['Maths', 'High', 'Yes', 'Yes']
]

S = data[0][:-1]
G = ['?', '?', '?']

print("S0 =", S)
print("G0 =", G)
print()

step = 1

for row in data:
    X = row[:-1]
    label = row[-1]

    print(f"I{step} =", X)

    if label == 'Yes':
    
        for i in range(len(S)):
            if S[i] != X[i]:
                S[i] = '?'

    else:
        
        for i in range(len(G)):
            if S[i] != X[i]:
                G[i] = S[i]

    print(f"S{step} =", S)
    print(f"G{step} =", G)
    print()

    step += 1

print("Final Specific Hypothesis =", S)
print("Final General Hypothesis =", G)
```
