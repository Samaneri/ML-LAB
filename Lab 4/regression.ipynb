```python
import matplotlib.pyplot as plt
import pandas as pd
from sklearn.linear_model import LinearRegression

data = {
    "week": [1, 2, 3, 4, 5],
    "sales": [1.2, 1.8, 2.6, 3.2, 3.8]
}

df = pd.DataFrame(data)

X = df[["week"]]
y = df["sales"]

model = LinearRegression()
model.fit(X, y)

predictions = model.predict(X)

print("Intercept:", model.intercept_)
print("Coefficient:", model.coef_[0])

print("\nRegression Equation:")
print(f"sales = {model.coef_[0]:.2f} * week + {model.intercept_:.2f}")

print(f"\nR² Score: {model.score(X, y):.4f}")

plt.figure(figsize=(6, 4))

plt.scatter(X, y, color="green", label="Actual Data")

plt.plot(X, predictions,
         color="blue",
         linewidth=2,
         label="Regression Line")

plt.title("Linear Regression: Week vs Sales")
plt.xlabel("Week")
plt.ylabel("Sales")

plt.legend()
plt.grid(True)
plt.show()
```

```python

import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
from sklearn.datasets import make_regression

X, y = make_regression(
    n_samples=50,
    n_features=1,
    noise=10,
    random_state=42
)

model = LinearRegression()
model.fit(X, y)

y_pred = model.predict(X)

print("Intercept:", model.intercept_)
print("Coefficient:", model.coef_[0])

plt.figure(figsize=(6, 4))

plt.scatter(X, y, color="green", label="Data points")

plt.plot(X, y_pred,
         color="red",
         linewidth=2,
         label="Regression Line")

plt.title("Linear Regression with Random Dataset")
plt.xlabel("X")
plt.ylabel("y")

plt.legend()
plt.grid(True)
plt.show()
```

```python

import numpy as np
from sklearn.decomposition import PCA

X = np.array([
    [2, 6],
    [1, 7]
])

print("Original Data:")
print(X)

pca = PCA(n_components=2)

X_pca = pca.fit_transform(X)

print("\nMean of Dataset:")
print(pca.mean_)

print("\nPrincipal Dataset (X_pca):")
print(X_pca)

X_reconstructed = pca.inverse_transform(X_pca)

print("\nReconstructed Matrix:")
print(X_reconstructed)

print("\nExplained Variance Ratio:")
print(pca.explained_variance_ratio_)

```

```python

import numpy as np

A = np.array([
    [1, 2],
    [3, 4],
    [5, 6]
])

print("Original Matrix (A):")
print(A)

U, S, VT = np.linalg.svd(A)

print("\nU Matrix:")
print(U)

print("\nSingular Values:")
print(S)

print("\nVᵀ Matrix:")
print(VT)

Sigma = np.zeros((3, 2))
Sigma[0, 0] = S[0]
Sigma[1, 1] = S[1]

print("\nSigma Matrix (Σ):")
print(Sigma)

A_reconstructed = U @ Sigma @ VT

print("\nReconstructed Matrix:")
print(A_reconstructed)

print("\nVerification (A == UΣVᵀ):")
print(np.allclose(A, A_reconstructed))

```

```python

import numpy as np
from sklearn.decomposition import PCA

X = np.array([
    [2, 6],
    [1, 7]
])

print("Original Data:")
print(X)

pca = PCA(n_components=2)
X_pca = pca.fit_transform(X)

print("\nMean of Dataset:")
print(pca.mean_)

print("\nPrincipal Dataset (X_pca):")
print(X_pca)

X_reconstructed = pca.inverse_transform(X_pca)

print("\nReconstructed Matrix:")
print(X_reconstructed)

print("\nExplained Variance Ratio:")
print(pca.explained_variance_ratio_)

X_centered = X - np.mean(X, axis=0)
cov_matrix = np.cov(X_centered.T)

print("\nCovariance Matrix:")
print(cov_matrix)

eigenvalues, eigenvectors = np.linalg.eig(cov_matrix)

print("\nEigen Values:")
print(eigenvalues)

print("\nEigen Vectors:")
print(eigenvectors)

```
