```python
import numpy as np
x = np.array([21, 22, 23, 24, 25])
y = np.array([31, 32, 33, 34, 35])
print("Array x:", x)
print("Array y:", y)
print("\nAddition (x + y):")
print(x + y)
print("\nSubtraction (x - y):")
print(x - y)
print("\nMultiplication (x * y):")
print(x * y)
print("\nDivision (x / y):")
print(x / y)
print("\nMean of x:")
print(np.mean(x))
print("\nMedian of x:")
print(np.median(x))
print("\nMaximum Value:")
print(np.max(x))
print("\nMinimum Value:")
print(np.min(x))
print("\nSum:")
print(x.sum())
print("\nVariance:")
print(x.var())
print("\nStandard Deviation:")
print(x.std())
```
