```python
import seaborn as sns
import pandas as pd
from matplotlib import pyplot as plt

data = {
    'week':[1,2,3,4,5],
    'sales':[1.2,1.8,2.6,3.2,3.8],
    'region':['south','south','north','north','north']
}

df = pd.DataFrame(data)



df1 = sns.load_dataset('iris')


sns.set(color_codes=True)



print("\nHistogram of week vs Sales\n")

sns.barplot(
    x="week",
    y="sales",
    hue="region",
    data=df
)

plt.title("Bar plot for Weeks Vs Sales Data")
plt.show()




print("\nBox Plot of week and sales data\n")

sns.boxplot(data=df)

plt.xlabel("Statistics", size=18)
plt.ylabel("Week and Sales Data", size=18)

plt.show()



sns.histplot(
    df1['sepal_width'],
    kde=False
)

plt.title("Distribution plot for sepal width")
plt.xlabel("Sepal Width")
plt.ylabel("Frequency")

plt.show()


# KDE Plot

sns.kdeplot(df1['sepal_width'])

plt.title("KDE Distribution plot for sepal width")
plt.xlabel("Sepal Width")
plt.ylabel("Density")

plt.show()




print("\nViolin Plot\n")

sns.violinplot(data=df)

plt.title("Violin Plot for Week and Sales Data")

plt.show()




sns.countplot(
    x="region",
    hue="region",
    data=df,
    palette="Blues",
    legend=False
)

plt.title("Count Plot Based on Regions")

plt.show()




print("\nStrip Plot\n")

sns.stripplot(
    x="species",
    y="petal_length",
    data=df1
)

plt.title("Strip Plot for Petal Length and Species")

plt.show()
```
