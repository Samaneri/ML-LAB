```python


import pandas as pd
import numpy as np
from math import log2
from sklearn.tree import DecisionTreeClassifier, export_text

data = {
    "Outlook": [
        "Sunny", "Sunny", "Overcast", "Rain", "Rain",
        "Rain", "Overcast", "Sunny", "Sunny", "Rain",
        "Sunny", "Overcast", "Overcast", "Rain"
    ],

    "Temperature": [
        "Hot", "Hot", "Hot", "Mild", "Cool",
        "Cool", "Cool", "Mild", "Cool", "Mild",
        "Mild", "Mild", "Hot", "Mild"
    ],

    "Humidity": [
        "High", "High", "High", "High", "Normal",
        "Normal", "Normal", "High", "Normal", "Normal",
        "Normal", "High", "Normal", "High"
    ],

    "Wind": [
        "Weak", "Strong", "Weak", "Weak", "Weak",
        "Strong", "Strong", "Weak", "Weak", "Weak",
        "Strong", "Strong", "Weak", "Strong"
    ],

    "PlayTennis": [
        "No", "No", "Yes", "Yes", "Yes",
        "No", "Yes", "No", "Yes", "Yes",
        "Yes", "Yes", "Yes", "No"
    ]
}

df = pd.DataFrame(data)

print("Dataset:")
print(df)


def entropy(target):

    values, counts = np.unique(target, return_counts=True)

    total = len(target)

    entropy_value = 0

    for count in counts:

        probability = count / total

        entropy_value -= probability * log2(probability)

    return entropy_value



def information_gain(data, feature, target):

  
    parent_entropy = entropy(data[target])

   
    values = data[feature].unique()

    weighted_entropy = 0

    for value in values:

        subset = data[data[feature] == value]

        probability = len(subset) / len(data)

        subset_entropy = entropy(subset[target])

        weighted_entropy += probability * subset_entropy

    gain = parent_entropy - weighted_entropy

    return gain


target = "PlayTennis"

print("\nEntropy of Dataset:")
print(entropy(df[target]))


features = [
    "Outlook",
    "Temperature",
    "Humidity",
    "Wind"
]

print("\nInformation Gain:")

for feature in features:

    gain = information_gain(
        df,
        feature,
        target
    )

    print(
        feature,
        "=",
        round(gain, 4)
    )



best_feature = None
best_gain = -1

for feature in features:

    gain = information_gain(
        df,
        feature,
        target
    )

    if gain > best_gain:

        best_gain = gain
        best_feature = feature


print("\nBest Attribute:")
print(best_feature)

print("\nHighest Information Gain:")
print(round(best_gain, 4))



X = pd.get_dummies(
    df[features],
    dtype=int
)

y = df[target].map({
    "No": 0,
    "Yes": 1
})



model = DecisionTreeClassifier(
    criterion="entropy",
    random_state=42
)



model.fit(X, y)



print("\nDecision Tree:")

tree = export_text(
    model,
    feature_names=list(X.columns)
)

print(tree)



new_data = pd.DataFrame([{
    "Outlook": "Overcast",
    "Temperature": "Mild",
    "Humidity": "High",
    "Wind": "Weak"
}])



new_data = pd.get_dummies(
    new_data,
    dtype=int
)

new_data = new_data.reindex(
    columns=X.columns,
    fill_value=0
)



prediction = model.predict(new_data)


print("\nPrediction:")

if prediction[0] == 1:
    print("Yes")
else:
    print("No")
```
