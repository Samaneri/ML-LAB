```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import classification_report, confusion_matrix
from sklearn.preprocessing import StandardScaler
data = {
    'CGPA': [9.2, 8.0, 8.5, 6.0, 6.5, 8.2, 5.8, 8.9],
    'Assessment': [85, 80, 81, 45, 50, 72, 38, 91],
    'Project': [8, 7, 8, 5, 4, 7, 5, 9],
    'Result': ['Pass', 'Pass', 'Pass', 'Fail',
               'Fail', 'Pass', 'Fail', 'Pass']
}
dataset = pd.DataFrame(data)
dataset.to_csv("knearest_csv.csv", index=False)
dataset = pd.read_csv("knearest_csv.csv")
print("Shape of Dataset:")
print(dataset.shape)
print("\nFirst 5 Records:")
print(dataset.head())
print("\nLast 2 Records:")
print(dataset.tail(2))
X = dataset.iloc[:, :-1].values
y = dataset.iloc[:, 3].values
print("\nInput Data (X):")
print(X)
print("\nOutput Data (y):")
print(y)
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=0
)
scaler = StandardScaler()

X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)
classifier = KNeighborsClassifier(n_neighbors=3)
classifier.fit(X_train, y_train)
y_pred = classifier.predict(X_test)
print("\nActual Values:")
print(y_test)
print("\nPredicted Values:")
print(y_pred)
print("\nConfusion Matrix:")
print(confusion_matrix(y_test, y_pred))
print("\nClassification Report:")
print(classification_report(y_test, y_pred))
new_student = [[9.1, 85, 8]]
new_student_scaled = scaler.transform(new_student)
predicted = classifier.predict(new_student_scaled)
print("\nNew Student:")
print(new_student)
print("\nPredicted Result:")
print(predicted)
```

```python
import pandas as pd
import numpy as np
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import classification_report, confusion_matrix, accuracy_score
import matplotlib.pyplot as plt
iris = load_iris()
X = iris.data
y = iris.target
target_names = iris.target_names
y_names = np.array([target_names[i] for i in y])
print("Shape of dataset:")
print(X.shape)
print("\nFirst 5 records:")
print(X[:5])
print("\nLast 2 records:")
print(X[-2:])
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y_names,
    test_size=0.20,
    random_state=42
)
classifier = KNeighborsClassifier(n_neighbors=3)
classifier.fit(X_train, y_train)
y_pred = classifier.predict(X_test)
print("\nConfusion Matrix:")
print(confusion_matrix(y_test, y_pred))
print("\nClassification Report:")
print(classification_report(y_test, y_pred))
accuracy = accuracy_score(y_test, y_pred) * 100
print("Accuracy of our model is equal "
      + str(round(accuracy, 2)) + " %.")
```
